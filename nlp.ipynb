```python
df=pd.read_csv('twitter_training.csv')
df
```

```python
df.rename(columns={list(df)[0]:'TweetID',
                   list(df)[1]:'Entity',
                      list(df)[2]:'Sentiment',
                      list(df)[3]:'Tweet'},inplace=True)
```

```python
df = df.drop(columns='TweetID')
```

```python
df.isnull().sum()
```

```python
import re
cleaned=[]
for txt in df['Tweet']:
    # Convert txt to string and handle NaN values
    txt = str(txt) if not pd.isna(txt) else ''
    cleaned_text = re.sub(r'[^\w\s]', '', txt)  # Removes everything except word characters and spaces
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text)  # Replaces multiple spaces with a single space
    cleaned_data=cleaned_text.strip()  # Removes leading/trailing whitespace
    cleaned.append(cleaned_data)
cleaned
```

```python
tokens=[]
for txt in cleaned:
  p=word_tokenize(txt)
  tokens.append(p)
tokens
```

```python
stopwords=stopwords.words('english')
stopwords
```

```python
filterd_txt=[]
for txt in tokens:
  for word in txt:
    if word not in stopwords:
      filterd_txt.append(word)
filterd_txt
```

```python
#stemming
stm=PorterStemmer()
stemmed_txt=[]
for word in filterd_txt:
  stemmed_txt.append(stm.stem(word))
stemmed_txt
```
