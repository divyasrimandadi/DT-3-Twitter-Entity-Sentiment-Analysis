```python
!pip install nltk
!pip install pandas
import nltk
import pandas as pd
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.stem import PorterStemmer
```

```python
df=pd.read_csv('twitter_training.csv')
df
```

```python
df.rename(columns={list(df)[0]:'TweetID',
                   list(df)[1]:'Entity',
                      list(df)[2]:'Sentiment',
                      list(df)[3]:'Tweet'},inplace=True)
```

```python
df.describe()
```

```python
df.info()
```

```python
df = df.drop(columns='TweetID')
```

```python
df.isnull().sum()
```

```python
import re
cleaned=[]
for txt in df['Tweet']:
    # Convert txt to string and handle NaN values
    txt = str(txt) if not pd.isna(txt) else ''
    cleaned_text = re.sub(r'[^\w\s]', '', txt)  # Removes everything except word characters and spaces
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text)  # Replaces multiple spaces with a single space
    cleaned_data=cleaned_text.strip()  # Removes leading/trailing whitespace
    cleaned.append(cleaned_data)
cleaned
```

```python
tokens=[]
for txt in cleaned:
  p=sent_tokenize(txt)
  tokens.append(p)
tokens
```

```python
stopwords=stopwords.words('english')
stopwords
```

```python
filterd_txt=[]
for txt in tokens:
  for word in txt:
    if word not in stopwords:
      filterd_txt.append(word)
filterd_txt
```

```python
#stemming
stm=PorterStemmer()
stemmed=[]
for word in filterd_txt:
  stemmed.append(stm.stem(word))
stemmed
```

```python
pip install scikit-learn
```

```python

import sklearn
from sklearn.feature_extraction.text import CountVectorizer
cv= CountVectorizer()
stem_ = [''.join(text) for text in stemmed]
stem_
```

```python
X=cv.fit_transform(stem_).toarray()
y=df['Sentiment']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```
